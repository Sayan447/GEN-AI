# Task 4 : Combined Operation

### 1.using the product_list and price_dict, create a list of tuples named catalog where each tuple is (product_name, price , category)

```python
product_list = [
    ("Laptop", "Electronics"),
    ("Mouse", "Electronics"),
    ("Keyboard", "Electronics"),
    ("Tablet", "Electronics"),
    ("Shoes", "Fashion"),
    ("Shirt", "Fashion")
]

price_dict = {
    "Laptop": 75000,
    "Mouse": 500,
    "Keyboard": 1200,
    "Tablet": 25000,
    "Shoes": 3000,
    "Shirt": 1500
}
```

```python
catalog = []

for product, category in product_list:
    price = price_dict.get(product, 0)
    catalog.append((product, price, category))

print("Catalog:", catalog)
```

### 2. from catalog, create a new dictionary category_to_products that maps each category to a list of product names in that category.

```python
category_to_products = {}

for product, price, category in catalog:
    if category not in category_to_products:
        category_to_products[category] = []
    
    category_to_products[category].append(product)

print("Category Mapping:", category_to_products)
```

## 3. Find category with max products & print them

```python
max_category = max(category_to_products, key=lambda k: len(category_to_products[k]))

print("Category with max products:", max_category)
print("Products in that category:", category_to_products[max_category])
```
