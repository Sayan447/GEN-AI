### pandas series is a column of data in a table like in excel or google sheets with labels called as indexes attached to each value

```python
pd.Series([40,50,60],index = ['Bread' , 'Milk' , 'Butter'],name ="Price of all items")
```

```python
# In a Pandas Series, the index acts like the key and the data acts like the value.
```

```python
# # Series from list
country = ['india' , 'china' , "USA" , 'UK']
pd.Series(country)
```

```python
batsman_runs=[48,64,72,98,114,183]
pd.Series(batsman_runs,dtype='int32')
```

## Series is one of the fundamental data structures in Pandas. It is used to store one-dimensional labeled data.

#### A Series lets you attach labels (indexes) to the values:

```python
subject_names = ['maths' , 'Physic' , 'bio' , 'Chemistry']
subject_marks = [48,92,64,87]

pd.Series(data = subject_marks, index=subject_names,name= "Subject Marks data", dtype = 'int32')
```

```python
### Creating an Series using a Numpy Array
values = np.arange(10,21)
pd.Series(values)
```

```python
# Dictionary to series
# Dictionary is another data structure that store the data  in key-value pairs and
# just because it stores the data in a key-value pairs so we don't have to explicitly mention the index
```

```python
data_items = {'Math' : 78 , "English" : 95 , "Physic" : 98}
pd.Series(data_items)
```

```python
# Custom index to series
values = [1,2,3,4]
pd.Series(data=values,index=['a','b','c','d'])
```

```python
# Scaler Value inside an series
pd.Series(4,index=['a','b','c','d'])
```
