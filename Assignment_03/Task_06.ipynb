```python
# Write a function process_prices(prices) that:
    # 1. Takes a list of prices.
    # 2. Uses map() + lambda to apply 10% discount to all prices.
    # 3. Uses filter() to keep only discounted prices that are above 300.
    # 4. return both lists:
    # discounted_prices
    # filtered_prices
# Test with:
# process_price([100,500,900,50,750])
```

```python
# Task 6: Complete process_prices function with map + filter
def process_prices(prices):
    """
    Takes a list of prices.
    Applies 10% discount using map() + lambda.
    Filters to keep only discounted prices above 300 using filter() + lambda.
    Returns both lists: discounted_prices and filtered_prices.
    """
    # Apply 10% discount
    discounted_prices = list(map(lambda price: price * 0.9, prices))
    
    # Keep only discounted prices above 300
    filtered_prices = list(filter(lambda price: price > 300, discounted_prices))
    
    return discounted_prices, filtered_prices

# Test the function
test_prices = [100, 500, 900, 50, 750]
discounted, filtered = process_prices(test_prices)

print("=" * 50)
print("TASK 6: Complete process_prices Function")
print("=" * 50)
print(f"Original Prices: {test_prices}")
print(f"Discounted Prices (10% off): {discounted}")
print(f"Filtered Prices (discounted > 300): {filtered}")
print()

# Detailed breakdown
print("Detailed Breakdown:")
print("-" * 50)
print(f"{'Original':<10} {'After 10%':<12} {'Status':<10}")
print("-" * 50)
for orig, disc in zip(test_prices, discounted):
    status = "KEPT" if disc > 300 else "REMOVED"
    print(f"{orig:<10} {disc:<12.2f} {status:<10}")
```
