## Combined Utility Function

```python
# Write a function process_prices(prices) that:
    # 1. Takes a list of prices.
    # 2. Uses map() + lambda to apply 10% discount to all prices.
    # 3. Uses filter() to keep only discounted prices that are above 300.
    # 4. return both lists:
    # discounted_prices
    # filtered_prices
# Test with:
# process_price([100,500,900,50,750])
```

```python
# Task 6: Combined Utility Function

def process_prices(prices):

    # Apply 10% discount
    discounted_prices = list(
        map(lambda price: price * 0.9, prices)
    )

    # Keep only prices above 300
    filtered_prices = list(
        filter(
            lambda price: price > 300,
            discounted_prices
        )
    )

    return discounted_prices, filtered_prices


# Test Case
discounted, filtered = process_prices(
    [100, 500, 900, 50, 750]
)

print("Discounted Prices:", discounted)
print("Filtered Prices:", filtered)
```
