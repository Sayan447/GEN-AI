## Task 2 -> Recursive Function: Factorial Utility

```python
# Create a recursive function factorial(n) that :
# 1. Return the factorial of n.
# 2. Handles two edge cases: n == 0 and n == 1.
# 3. Print an error message if n is negative.
# Test With:-----
    # factorial(5)
    # factorial(0)
    # factorial(-3)
```

```python
# Task 2: Recursive Function - Factorial

def factorial(n):

    # Negative number check
    if n < 0:
        print("Error: Factorial is not defined for negative numbers")
        return None

    # Base cases
    if n == 0 or n == 1:
        return 1

    # Recursive case
    return n * factorial(n - 1)


# Test Cases
print(factorial(5))
print(factorial(0))
print(factorial(-3))
```
