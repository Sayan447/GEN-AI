## Task 4 - Using map():Apply GST to list of Prices

```python
# Given:
# prices = [100 , 250 ,400, 1200 , 50]
# Use map() with your gst lambda to generate a new list price_with_gst.
# Print both lists: Original Prices, Prices after GST
```

```python
# Task 4: Using map() to Apply GST

prices = [100, 250, 400, 1200, 50]

gst = lambda price: price + (price * 0.18)

price_with_gst = list(map(gst, prices))

print("Original Prices:", prices)
print("Prices After GST:", price_with_gst)
```
