## Task 3- Lambda Function: GST calculator

```python
# create a lambda function gst that return price after adding 18% GST
# Example : 
# gst = lambda price: price + (0.18 * price)
# print(gst(100)) #should return 118
# Extra --> Create another lambda to compute final price after the GST + discount together
```

```python
# Task 3: Lambda functions for GST
# Basic GST lambda (adds 18% GST)
gst = lambda price: price * 1.18

# Combined GST + discount (GST applied first, then discount)
final_price = lambda price, discount: price * 1.18 * (1 - discount/100)

print("=" * 50)
print("TASK 3: Lambda Functions for GST")
print("=" * 50)
print(f"gst(100): {gst(100)}")
print(f"final_price(100, 10): {final_price(100, 10)}")
print()
```
