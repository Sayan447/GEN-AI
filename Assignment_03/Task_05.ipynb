## Using filter(): filter Expensive Products

```python
# Given the list:
    # prices = [100,250,400,1200,50,2000,850]
    # use filter() to:
    # 1.Create a list of prices greater than 500.
    # 2.Create another list of prices less than or equal to 500.
# print both lists.
```

```python
# Task 5: Using filter() to separate prices
prices_list = [100, 250, 400, 1200, 50, 2000, 850]

# Create filtered lists
greater_than_500 = list(filter(lambda price: price > 500, prices_list))
less_or_equal_500 = list(filter(lambda price: price <= 500, prices_list))

print("=" * 50)
print("TASK 5: Using filter() for Price Ranges")
print("=" * 50)
print(f"Original Prices: {prices_list}")
print(f"Prices greater than 500: {greater_than_500}")
print(f"Prices less than or equal to 500: {less_or_equal_500}")
print()
```
