## Using filter(): filter Expensive Products

```python
# Given the list:
    # prices = [100,250,400,1200,50,2000,850]
    # use filter() to:
    # 1.Create a list of prices greater than 500.
    # 2.Create another list of prices less than or equal to 500.
# print both lists.
```

```python
# Given list
prices = [100, 250, 400, 1200, 50, 2000, 850]

# Prices greater than 500
greater_than_500 = list(filter(lambda x: x > 500, prices))

# Prices less than or equal to 500
less_or_equal_500 = list(filter(lambda x: x <= 500, prices))

# Print results
print("Prices > 500:", greater_than_500)
print("Prices <= 500:", less_or_equal_500)
```
