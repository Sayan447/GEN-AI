### Basic Function: Price After Discount

```python
# Write a function apply_discount(price , discount_percent) that:
# 1.Return the price after discount,
# 2.If discount_percent is missing, apply a default discount of 5%.
# 3. Test the fucntion with:
# 4. apply_discount(1000 , 10)
# 5. apply_discount(500) #uses default discount
# ----Extra(optional): Add a condition inside the fucntion to ensure discount never  exceeds 60%.
```

```python
# Task 1: apply_discount function
def apply_discount(price, discount_percent=None):
    """
    Return price after discount.
    Default discount is 5% if discount_percent is missing.
    Discount never exceeds 60%.
    """
    if discount_percent is None:
        discount_percent = 5
    
    # Ensure discount never exceeds 60%
    if discount_percent > 60:
        discount_percent = 60
    
    return price * (100 - discount_percent) / 100

# Test cases
print("=" * 50)
print("TASK 1: Apply Discount Function")
print("=" * 50)
print(f"apply_discount(1000, 10): {apply_discount(1000, 10)}")
print(f"apply_discount(500): {apply_discount(500)}")
print(f"apply_discount(1000, 80): {apply_discount(1000, 80)} (capped at 60%)")
print()
```
