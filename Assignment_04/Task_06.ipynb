## Task 6: Read File Safely(Error Handling Inside File Handling Only) 

```python
# You Must not use exceptions beyond file-related safeguards here.
# 1. Ask the user for a filename to open.
# 2. If the file exist, read and print it.
# 3. If it does not exist, print:
# "File not found. Please check the filename."
# Use simple condition checks with os.path.exists() [allowed]
```

```python
import os

# 1. Ask the user for a filename to open
print("=" * 40)
print("     FILE READER")
print("=" * 40)

filename = input("Enter the filename to open: ")

# 2. & 3. Check if file exists using os.path.exists()
if os.path.exists(filename):
    # File exists - read and print it
    with open(filename, 'r') as file:
        content = file.read()
        print("\n" + "=" * 40)
        print(f"     CONTENTS OF '{filename}'")
        print("=" * 40)
        print(content)
        print("=" * 40)
else:
    # File does not exist - print error message
    print("\n File not found. Please check the filename.")
```
