## Task 2: Read file in Different ways

```python
# Using the same sales_data.txt:
# 1. read the entire file using .read() and print it.
# 2. Read the first line using .readline().
# 3. Read all line using .readlines() and convert them into a list of integers.
# Ensure proper formatting and cleanup of newline characters. 
```

```python
# ==========================================
# Task 2: Read File in Different Ways
# ==========================================

# 1. Read full file
with open("sales_data.txt", "r") as file:

    full_content = file.read()

print("Full File Content:")
print(full_content)


# 2. Read first line
with open("sales_data.txt", "r") as file:

    first_line = file.readline().strip()

print("First Line:")
print(first_line)


# 3. Read all lines and convert to integers
with open("sales_data.txt", "r") as file:

    lines = file.readlines()

sales_list = []

for line in lines:

    cleaned_line = line.strip()

    if cleaned_line:
        sales_list.append(int(cleaned_line))

print("Sales List:")
print(sales_list)
```
