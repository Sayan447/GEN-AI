## Task 2: Read file in Different ways

```python
# Using the same sales_data.txt:
# 1. read the entire file using .read() and print it.
# 2. Read the first line using .readline().
# 3. Read all line using .readlines() and convert them into a list of integers.
# Ensure proper formatting and cleanup of newline characters. 
```

```python
with open("sales_data.txt", "r") as file:
    content = file.read()
    print("Full file content:")
    print(content)
```

```python
with open("sales_data.txt", "r") as file:
    first_line = file.readline().strip()   # remove newline
    print("First line:")
    print(first_line)
```

```python
with open("sales_data.txt", "r") as file:
    lines = file.readlines()

# Clean and convert to integers
sales_list = [int(line.strip()) for line in lines]

print("Sales as list of integers:")
print(sales_list)
```
