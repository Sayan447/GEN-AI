## Append New Sales:


```python
# 1 Append these new sales to the same file:
# 5000 , 2500 , 1700
# 2. After appending , reopen and print the entire update file
# Extra (optional) : Print the total number of lines after appending
```

```python
# ==========================================
# Task 3: Append New Sales
# ==========================================

new_sales = [5000, 2500, 1700]

# Append sales
with open("sales_data.txt", "a") as file:

    for sale in new_sales:
        file.write(str(sale) + "\n")

print("New sales added successfully!")


# Read updated file
with open("sales_data.txt", "r") as file:

    content = file.read()

print("\nUpdated File Content:")
print(content)


# Extra: Count total lines
with open("sales_data.txt", "r") as file:

    total_lines = len(file.readlines())

print("Total Lines:", total_lines)
```
