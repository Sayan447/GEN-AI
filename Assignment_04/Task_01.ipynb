## Task 1: Write Sales Records to a file

```python
# 1.Create a list of sales amounts:
# 2.sales = [1200 ,450 , 980 , 1500 , 3000]
# 3. Open a file named sales_data.txt in write mode.
# 4. write each sales on a new line in the file.
# 5. Close the file, then reopen it and print its contents.
# Extra : Write the data in comma-separated format instead of separate lines.
```

```python
# 1. Create a list of sales amounts
sales = [1200, 450, 980, 1500, 3000]

# 2. Open a file in write mode
file = open("sales_data.txt", "w")

# 3. Write each sale on a new line
for s in sales:
    file.write(str(s) + "\n")

# 4. Close the file
file.close()

# 5. Reopen the file and print contents
file = open("sales_data.txt", "r")
content = file.read()
print("File contents (line by line):")
print(content)

file.close()
```
