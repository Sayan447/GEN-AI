## Generate summery report from file

```python
# Using only file reading operations:
# 1. Read all sales values from sale_data.txt.
# 2. Convert them into integers.
# 3. Calculate and print:
#   * total sales
#   * Higher sales
#   * Lower sales
#   * Average sales
# Do not use any advanced libraries
```

```python
 

# 1. Read all sales values from sales_data.txt
with open('sales_data.txt', 'r') as file:
    # Read all lines using .readlines()
    lines = file.readlines()

# 2. Convert them into integers
sales_values = []
for line in lines:
    # Strip newline characters and convert to integer
    # Skip empty lines if any
    if line.strip():  # Check if line is not empty
        sales_values.append(int(line.strip()))

# 3. Calculate and print statistics
if sales_values:  # Ensure list is not empty
    total_sales = 0
    highest_sales = sales_values[0]
    lowest_sales = sales_values[0]
    
    # Calculate total, highest, and lowest
    for value in sales_values:
        total_sales += value
        if value > highest_sales:
            highest_sales = value
        if value < lowest_sales:
            lowest_sales = value
    
    # Calculate average
    average_sales = total_sales / len(sales_values)
    
    # Print results
    print("=== Sales Statistics ===")
    print(f"Total sales: ${total_sales:,}")
    print(f"Highest sales: ${highest_sales:,}")
    print(f"Lowest sales: ${lowest_sales:,}")
    print(f"Average sales: ${average_sales:.2f}")
    print(f"Number of sales records: {len(sales_values)}")
else:
    print("No sales data found in the file.")
```
