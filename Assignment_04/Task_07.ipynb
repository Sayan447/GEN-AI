## Task 7 : Mini Project - Export Discounted Prices

```python
# Create a dictionary:
# prices = {
    # "Mouse" : 500,
    # "Keyboard" : 800,
    #  "Monitor" : 7000,
    #  "Pendrive" : 400,
    # "Camera" : 5000 
# }
# Ask the user for a discount percentage.
# Write discounted prices into disocunt_report.txt using:
#  Product | Original Price | Discounted Price
# After writing, read the file and print it to the terminal.
# Extra : Write a summery at the bottom of the file.
# Total Items: X 
# Average Discountd Price : Y
```

```python
# Create the dictionary
prices = {
    "Mouse": 500,
    "Keyboard": 800,
    "Monitor": 7000,
    "Pendrive": 400,
    "Camera": 5000
}

# Ask the user for a discount percentage
print("=" * 60)
print("     DISCOUNT CALCULATOR")
print("=" * 60)

try:
    discount_percent = float(input("Enter discount percentage (e.g., 10 for 10%): "))
    
    # Calculate discount factor
    discount_factor = (100 - discount_percent) / 100
    
    # Write discounted prices into discount_report.txt
    with open('discount_report.txt', 'w') as file:
        # Write header
        file.write("=" * 60 + "\n")
        file.write(f"{'Product':<15} | {'Original Price':>15} | {'Discounted Price':>18}\n")
        file.write("=" * 60 + "\n")
        
        # Calculate total for average
        total_discounted = 0
        total_items = 0
        
        # Write each product's data
        for product, original_price in prices.items():
            discounted_price = original_price * discount_factor
            total_discounted += discounted_price
            total_items += 1
            
            # Format with proper spacing
            file.write(f"{product:<15} | ${original_price:>14.2f} | ${discounted_price:>17.2f}\n")
        
        file.write("=" * 60 + "\n")
        
        # Extra: Write summary at the bottom of the file
        average_discounted = total_discounted / total_items
        file.write(f"\n{'=' * 60}\n")
        file.write(f"SUMMARY:\n")
        file.write(f"Total Items: {total_items}\n")
        file.write(f"Average Discounted Price: ${average_discounted:.2f}\n")
        file.write(f"Discount Applied: {discount_percent}%\n")
        file.write("=" * 60 + "\n")
    
    print(f"\n Discount report saved to 'discount_report.txt' with {discount_percent}% discount!")
    
    # After writing, read the file and print it to the terminal
    print("\n" + "=" * 60)
    print("     DISCOUNT REPORT")
    print("=" * 60)
    
    with open('discount_report.txt', 'r') as file:
        content = file.read()
        print(content)
    
except ValueError:
    print(" Invalid input! Please enter a valid number for discount percentage.")
```
