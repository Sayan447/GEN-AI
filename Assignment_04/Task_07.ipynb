## Task 7 : Mini Project - Export Discounted Prices

```python
# Create a dictionary:
# prices = {
    # "Mouse" : 500,
    # "Keyboard" : 800,
    #  "Monitor" : 7000,
    #  "Pendrive" : 400,
    # "Camera" : 5000 
# }
# Ask the user for a discount percentage.
# Write discounted prices into disocunt_report.txt using:
#  Product | Original Price | Discounted Price
# After writing, read the file and print it to the terminal.
# Extra : Write a summery at the bottom of the file.
# Total Items: X 
# Average Discountd Price : Y
```

```python
# ==========================================
# Task 7: Export Discounted Prices
# ==========================================

prices = {
    "Mouse": 500,
    "Keyboard": 800,
    "Monitor": 7000,
    "Pendrive": 400,
    "Camera": 5000
}

try:

    discount_percent = float(
        input(
            "Enter discount percentage: "
        )
    )

    total_discounted = 0
    total_items = len(prices)

    with open(
        "discount_report.txt",
        "w"
    ) as file:

        file.write(
            "Product | "
            "Original Price | "
            "Discounted Price\n"
        )

        file.write(
            "-" * 50 + "\n"
        )

        for product, price in prices.items():

            discounted_price = (
                price -
                (
                    price *
                    discount_percent /
                    100
                )
            )

            total_discounted += (
                discounted_price
            )

            file.write(
                f"{product} | "
                f"{price} | "
                f"{discounted_price:.2f}\n"
            )

        # Summary
        average_discounted = (
            total_discounted /
            total_items
        )

        file.write("\nSUMMARY\n")

        file.write(
            f"Total Items: "
            f"{total_items}\n"
        )

        file.write(
            f"Average Discounted Price: "
            f"{average_discounted:.2f}\n"
        )

    # Read and print report
    with open(
        "discount_report.txt",
        "r"
    ) as file:

        print("\nDiscount Report:")
        print(file.read())

except ValueError:

    print(
        "Invalid input! "
        "Enter a number."
    )
```
