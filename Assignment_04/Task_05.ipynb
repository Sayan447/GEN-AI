## Create Product info file(user input)

```python
# 1 . Ask the user for 3 product names & their prices.
# 2 . Write them into a new products.txt in this format.
# 3 . ProductName | Price.
# 4 . Read the file and print each line with proper formatting.
```

```python
# ==========================================
# Task 5: Create Product Info File
# ==========================================

products = []

# Get user input
for i in range(3):

    print(f"\nEnter Product {i+1}")

    product_name = input(
        "Enter product name: "
    )

    product_price = float(
        input(
            "Enter product price: "
        )
    )

    products.append(
        (
            product_name,
            product_price
        )
    )


# Write to file
with open("products.txt", "w") as file:

    for name, price in products:

        file.write(
            f"{name} | {price}\n"
        )

print("Products saved successfully!")


# Read and display
print("\nProduct List:")

with open("products.txt", "r") as file:

    for line in file:
        print(line.strip())
```
