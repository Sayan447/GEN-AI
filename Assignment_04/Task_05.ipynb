## Create Product info file(user input)

```python
# 1 . Ask the user for 3 product names & their prices.
# 2 . Write them into a new products.txt in this format.
# 3 . ProductName | Price.
# 4 . Read the file and print each line with proper formatting.
```

```python
# 1. Ask the user for 3 product names & their prices
print("=" * 40)
print("     PRODUCT ENTRY SYSTEM")
print("=" * 40)

products = []
for i in range(3):
    print(f"\nProduct {i+1}:")
    name = input("  Product name: ")
    price = input("  Product price: $")
    products.append((name, price))

# 2. Write them into a new products.txt in the format: ProductName | Price
with open('products.txt', 'w') as file:
    for name, price in products:
        file.write(f"{name} | {price}\n")

print("\n Products have been saved to 'products.txt'!")

# 3. Read the file and print each line with proper formatting
print("\n" + "=" * 40)
print("     PRODUCT LIST")
print("=" * 40)

with open('products.txt', 'r') as file:
    lines = file.readlines()
    for line in lines:
        # Print with proper formatting, removing the newline character
        print(f"  • {line.strip()}")

print("=" * 40)
```
