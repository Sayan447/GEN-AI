## Task 2: Process Multiple Orders(for loop)


```python
# given a list of order amounts: orders = [1200,2500,800,1750,3000], use for loop
#  to apply the discount rules from Task 1 to each order and print a summary table showing:
# order_amount--> discount% --> final_amount.
```

```python
orders = [1200, 2500, 800, 1750, 3000]

print("Order Amount --> Discount % --> Final Amount")

for order_amount in orders:
    # Apply discount rules
    if order_amount >= 2000:
        discount = 0.15
    elif order_amount >= 1500:
        discount = 0.10
    elif order_amount >= 1000:
        discount = 0.07
    else:
        discount = 0

    # Calculate final amount
    final_amount = int(order_amount - (order_amount * discount))

    # Print result
    print(f"{order_amount} --> {discount*100}% --> {final_amount}")
```

```python
# # Also compute and print the total revenue after discount.
# Extra (optional) : Print the number of orders that received a discount (discount > 0)
```

```python
orders = [1200, 2500, 800, 1750, 3000]

total_revenue = 0
discounted_orders = 0

print("Order Amount --> Discount % --> Final Amount")

for order_amount in orders:
    # Apply discount rules
    if order_amount >= 2000:
        discount = 0.15
    elif order_amount >= 1500:
        discount = 0.10
    elif order_amount >= 1000:
        discount = 0.07
    else:
        discount = 0

    # Count discounted orders
    if discount > 0:
        discounted_orders += 1

    # Calculate final amount
    final_amount = int(order_amount - (order_amount * discount)
)
    # Add to total revenue
    total_revenue += final_amount

    # Print row
    print(f"{order_amount} --> {discount*100}% --> {final_amount}")

# Print totals
print("\nTotal Revenue after Discount:", total_revenue)
print("Number of Discounted Orders:", discounted_orders)
```
