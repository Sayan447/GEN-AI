```python
## apply the following:
# order_amount >= 2000--> 15% discount
# 1500 <= order_amount < 2000 --> 10% discount
# 1000 <= order_amount < 1500 --> 7% discount
# otherwise --> 0% discount
```

```python
# Read order amount from user
order_amount = int(input("Enter the order amount: "))

# Apply discount based on conditions
if order_amount >= 2000:
    discount = 0.15
elif order_amount >= 1500:
    discount = 0.10
elif order_amount >= 1000:
    discount = 0.07
else:
    discount = 0

# Calculate discount amount and final price
discount_amount = order_amount * discount
final_amount = order_amount - discount_amount

# Print results
print("Original Amount:", order_amount)
print("Discount Applied:", discount * 100, "%")
print("Discount Amount:", int(discount_amount))
print("Final Amount to Pay:", int(final_amount))
```

```python
# 3. Ensure you convert input to a number and handle the case when the input is non- numeric by displaying an error message and exiting the program.

# Extra(optional): Add tax(fixed 5%) after discount and print subtotal,tax,and final total(still within conditional and basic arithmetic)
```

```python
# Read input safely
try:
    order_amount = float(input("Enter the order amount: "))
except ValueError:
    print("Error: Please enter a valid numeric value.")
    exit()

# Apply discount
if order_amount >= 2000:
    discount = 0.15
elif order_amount >= 1500:
    discount = 0.10
elif order_amount >= 1000:
    discount = 0.07
else:
    discount = 0

# Calculate subtotal after discount
discount_amount = order_amount * discount
subtotal = order_amount - discount_amount

# Apply 5% tax on subtotal
tax = subtotal * 0.05

# Final total
final_total = subtotal + tax

# Print results
print("Original Amount:", order_amount)
print("Discount Applied:", discount * 100, "%")
print("Discount Amount:", int(discount_amount))
print("Subtotal after Discount:", int(subtotal))
print("Tax (5%):", tax)
print("Final Total:", int(final_total))
```
