## Task 4 : Loop control with conditions (break & continue)

```python
# 1. Given a list of daily sales: daily = [200 ,150 , 0 ,400 , 50 , -1,300], iterate through it with a for loop and : i) if a day's value is-1, treat it as corrupted data and break the loop (stop processing) ii) if a day's value is 0, treat it as a day with no sales and continue ( skip adding to revenue). iii) for valid positive sales, add to total_sales and print the running total.
```

```python
daily = [200, 150, 0, 400, 50, -1, 300]

total_sales = 0

for sale in daily:
    # Check for corrupted data
    if sale == -1:
        print("Corrupted data found. Stopping processing.")
        break

    # Skip days with no sales
    if sale == 0:
        print("No sales today. Skipping.")
        continue

    # Add valid sales
    total_sales += sale
    print("Added:", sale, "| Running Total:", total_sales)

print("Final Total Sales:", total_sales)
```

```python
# 2. print the final total after the loop complete( or stops due to corrupted data)
```

```python
daily = [200, 150, 0, 400, 50, -1, 300]

total_sales = 0

for sale in daily:
    if sale == -1:
        print("Corrupted data found. Stopping processing.")
        break

    elif sale == 0:
        print("No sales today. Skipping.")
        continue

    else:
        total_sales += sale
        print("Added:", sale, "| Running Total:", total_sales)

# Final total printed AFTER loop ends (normal or break)
print("Final Total Sales:", total_sales)
```
