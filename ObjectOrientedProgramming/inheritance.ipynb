```python
class Avenger:
    
    def __init__(self, name , power):
        self.name = name
        self.power=power


    def introduction(self):
        print(f"hello my name is {self.name}")
    
    
class Ironman(Avenger):
    def __init__(self, name, power, weapon):
        super().__init__(name,power)
        self.weapon=weapon
        
    def weapon_details(self):
        print(f"I use {self.weapon}")
    
```

```python
iron=Ironman("Ironman" , "Genius AI","Beam");
```

```python
iron.name
```

```python
iron.weapon
```

```python
iron.weapon_details()
```

```python
iron.introduction()
```

#### Multilevel Inheritance ---> The properties or Methods and attributes of the grandparent class can be inherited by the parent  class and then the properties of the  parent class will be inherited by the child class.

```python
class Grandparent:
    
    def __init__(self, nationality , baldness,anger_issues):
        self.nationality=nationality
        self.baldness=baldness
        self.anger_issues=anger_issues
        
        
    def introduction(self):
        print(f"hello my nationality is : {self.nationality} and about my anger_issues if you want to ask about then the answer is {self.anger_issues}")
```

```python
class Parent(Grandparent):
    def __init__(self, nationality , baldness,anger_issues, intelligence):
        super().__init__(nationality,baldness,anger_issues)
        self.intelligence = intelligence
        
        
        
    def intelligence_details(self):
        print(f"{self.intelligence}")
        
        
    
```

```python
class child(Parent):
    def __init__(self, nationality , baldness,anger_issues, intelligence,charm):
        super().__init__(nationality,baldness,anger_issues)
        self.intelligence = intelligence
        
        
        
    def charm_details(self):
        print(f"{self.charm}")
        
        
    
```

```python
parent_1=Parent('Indian' , "Yes" , "No" , "Yes")
```

```python
parent_1.introduction()
```
