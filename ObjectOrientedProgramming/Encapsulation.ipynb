# Encapsulation

#### Encapsulation is a way of wrapping data(Variables and methods) inside a capsule or a class as a single unit 

##### it restrict the direct access to same of the object components which is a means of preventing accidental inference and misuse of tha e data

## key concept of encapsulation
## Public Members
## Private Members
## Protected Members
## getter and setter methods

```python
class Ironman:
    def __init__(self,name,power,weapon):
        self.name=name
        self.power=power
        self.weapon=weapon
        
        
    def introduction(self):
        print(f"Hello my name is{self.name} and i have {self.weapon}")
```

```python
iron=Ironman("amit" , "Gen AI" , "Beam")
```

##### dir()---> It takes the input as the object name and gives us this idea that which member is public or protected or private

```python
dir(iron)
```

```python
iron.power
```

```python
iron.weapon
```

```python
iron.introduction()
```

```python
class Avenger:
    def __init__(self,name,phone_number):
        self.name = name
        self.__phone_number=phone_number
        
# The double underscore __ makes it private (name mangling).
```

```python
avenger=Avenger("captain" , "742355")
```

```python
avenger.name
```

```python
avenger.phone_number
# no attributes because the it is coming from the private class 
```

```python
dir(avenger)
```

```python
from operator import add


class Captain:
    def __init__(self, name ,address):
        self.name=name
        self.__address = address
    
    
    # getter method
    def accessing_add(self):
        print(f'My address is {self.__address}')
```

```python
captain=Captain('captain america' , 'new York')
```

```python
captain.name
```

```python
captain.address
```

```python
dir(captain)
```

```python
captain.accessing_add()
```

```python
class Animal:
    def __init__(self,name,sound):
        self.name=name
        self.__sound=sound
        
    def sound_info(self):
        print(f"the animal is going to sound as {self.__sound}")
        
        
    def sound_modify(self,new_sound):
        self.__sound=new_sound
        print(f"New sound for the animal is : {self.__sound}")
```

```python
animal =Animal("Bruno" , "woof woof")
```

```python
animal.sound_info()
```

```python
animal.sound
```

```python
# getter method to have the access over the data 
```

```python
animal.sound_modify("meaw meaw")
```

```python
class BankAccount:

    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.__balance = balance

    def get_balance(self):
        return self.__balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            return f"Deposited amount: {amount}. Total balance: {self.get_balance()}"
        else:
            return "Deposit amount must be positive"

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            return f"Withdrawn amount: {amount}. Remaining balance: {self.get_balance()}"
        else:
            return "Insufficient funds in the account"


# Creating object
# acc1 = BankAccount("Boss", 5000)

# print(acc1.deposit(1000))
# print(acc1.withdraw(2000))
# print(acc1.get_balance())
```
