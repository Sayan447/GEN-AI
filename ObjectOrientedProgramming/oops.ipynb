```python
class Cartoon:
    
    def __init__(self,name,show,catchphrase):
        self.name = name
        self.show = show
        self.catchphrase = catchphrase
        
    def introdunction(self):
        print(f"Hello my  name is {self.name}")
        
        
        
    def speak(self):
        print(f"{self.name} says {self.catchphrase}")
```

## default Constructor : It doesnot  ahve the parameters inside it and it is only having the self keyword inside it. it assigns a default value to all of the object

```python
class Dog:
    
    def __init__(self):
        self.name = "Labrador"
        self.color = "Brown"
        
    def show_info(self):
        print(f"this dog is {self.name} and color of the dog is {self.color}")
```

```python
dog1 = Dog()
dog1.name
```

```python
dog1.show_info()
```

## Parameterized Constructor: - it consist of parameter inside it so that every object can have a specific attribute associate to it

### types of Method
### 1 . Instance Method
### 2 . Static Method
### 3 . Class Method

```python
# The Instance methods is going to use self keyword to access and modify object attribute
```

```python
class Avengerss:
    
    def __init__(self,name,power):
        self.name = name
        self.power = power
    
    
    @staticmethod
    def mission():
        print("All Avengers must protect the earth")
        
        
    origin = "Avenger"
    
    @classmethod
    def origin_info(cls):
        print(f"The origin for the Avenger is: {cls.origin}")
        
    def introduction(self):
        print(f"My name is {self.name} and power is {self.power}")
```

```python
Iron_man=Avengerss("iron man" , "Gen AI")
```

```python
Iron_man.introduction()
```

```python
Iron_man.origin_info()
```

```python
Iron_man.mission()
```
