## Magic Method

#### Magic methods are also called as Dunder Methoda because they have double underscore __before and after their name and they are special methods in python that allows us to customized how objects behave

#### these methods can help us define:
##### 1. Object Intialization (__init__)
##### 2. String Representation (__str__,__repr__)
##### 3. Arithmetic Operations(__add__,__sub__,etc)
##### 4. Comparison Operations(__eq__,__lt__)
##### 4. Length, indexing, iteration(__len__,__getitem__,etc)


###### Why we use:
#### 1. Makes objects behave like built in types(eg, list, integers)
#### 2. Improves readability by allowing basic operation object1+object2
#### 3. Simplification code by replacing verbose function calls with natural syntax

```python
# print(object) --> str --> what is returned when print(object) is used
```

##### arithmetic operations can be done among the objects by using magic methods

```python
class Avenger :
    
    def __init__(self,name,strength):
        self.name=name
        self.strength=strength
        
        
    def __add__(self,other):
        return Avenger(self.name + "&" + other.name ,self.strength + other.strength)
    
    def __str__(self):
        return f"{self.name} with combined strength {self.strength}"
        
        
ironman = Avenger("Ironman" , 80)
thor=Avenger("Thor" , 100)

team = ironman + thor
print(team)
```
