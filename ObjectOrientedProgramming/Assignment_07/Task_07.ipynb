## Mini Project: Simple Inventory System(OOPs Only)

```python
# Create two classes:
# class : Invertory
# Attribute :
# product --> list to store product object
# methods: add_product(product), remove_product(name),get_total_value()--> sums prices of all product
# show_all_products() --> prints info for each product
# Class: Store
# Attributes: store_name
#           inventory --> an inventory object
# Methods:
    # add_new_product()--> takes input & create product object
    # show_summery() --> prints total items & value
    
# important: Use oly OOP concept -- no file handling, no file handling, no exceptions, no packeages.

# Test the system by:
# 1. creating a store object
# 2. adding 3 products 
# 3. showing summery
# 4.using __add__ to combine prices of two products
```

```python
# Product Class
class Product:

    # Constructor
    def __init__(self, name, price, category):
        self.name = name
        self.__price = price
        self.category = category

    # Getter for price
    def get_price(self):
        return self.__price

    # Setter for price
    def set_price(self, new_price):
        if new_price > 0:
            self.__price = new_price

    # Product details
    def get_info(self):
        print(f"Name: {self.name}")
        print(f"Price: {self.__price}")
        print(f"Category: {self.category}")

    # Magic Method (__str__)
    def __str__(self):
        return f"Product({self.name}, {self.__price}, {self.category})"

    # Operator Overloading (__add__)
    def __add__(self, other):
        return self.__price + other.get_price()


# Inventory Class
class Inventory:

    # Constructor
    def __init__(self):
        self.products = []   # list to store products

    # Add product
    def add_product(self, product):
        self.products.append(product)

    # Remove product by name
    def remove_product(self, name):
        for product in self.products:
            if product.name == name:
                self.products.remove(product)
                print(name, "removed successfully")

    # Total value of all products
    def get_total_value(self):
        total = 0
        for product in self.products:
            total += product.get_price()
        return total

    # Show all products
    def show_all_products(self):
        print("\nAll Products:")
        for product in self.products:
            product.get_info()
            print()


# Store Class
class Store:

    # Constructor
    def __init__(self, store_name):
        self.store_name = store_name
        self.inventory = Inventory()

    # Add new product
    def add_new_product(self, name, price, category):
        product = Product(name, price, category)
        self.inventory.add_product(product)

    # Show summary
    def show_summary(self):
        print("\nStore Name:", self.store_name)
        print("Total Items:", len(self.inventory.products))
        print("Total Inventory Value:", self.inventory.get_total_value())


# ---------------- TEST SYSTEM ----------------

# 1. Create store object
store = Store("Tech Store")

# 2. Add 3 products
store.add_new_product("Laptop", 50000, "Electronics")
store.add_new_product("Mobile", 25000, "Electronics")
store.add_new_product("Headphones", 3000, "Accessories")

# Show all products
store.inventory.show_all_products()

# 3. Show summary
store.show_summary()

# 4. Use __add__ to combine prices of two products
product1 = store.inventory.products[0]
product2 = store.inventory.products[1]

total_price = product1 + product2

print("\nCombined Price of Two Products:", total_price)
```
