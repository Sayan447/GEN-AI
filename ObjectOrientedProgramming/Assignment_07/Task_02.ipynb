## Cosntructor & Encapsulation

```python
# Modify the product class: 
# Make price a private attribute(__price).
# Create getter & setter methods:
#   get_price()
#   set_price(new_price)--> should update only if new_price > 0
# test modifying price using the setter
```

```python
# Create Product class
class Product:

    # Constructor
    def __init__(self, name, price, category):
        self.name = name
        self.__price = price   # Private attribute
        self.category = category

    # Getter method
    def get_price(self):
        return self.__price

    # Setter method
    def set_price(self, new_price):
        if new_price > 0:
            self.__price = new_price
            print("Price updated successfully")
        else:
            print("Price must be greater than 0")

    # Method to print product details
    def get_info(self):
        print("Product Name:", self.name)
        print("Price:", self.__price)
        print("Category:", self.category)


# Create object
product1 = Product("Laptop", 50000, "Electronics")

# Show original details
print("Original Product Details:")
product1.get_info()

# Get price using getter
print("\nCurrent Price:", product1.get_price())

# Modify price using setter
product1.set_price(45000)

# Check updated price
print("\nUpdated Product Details:")
product1.get_info()

# Test invalid price
product1.set_price(-1000)
```
