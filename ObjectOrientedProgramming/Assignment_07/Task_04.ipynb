## Polymorphism

```python
# create two classes:
    # laptop(Product)
    # Mobile(Product)
# both override:
# def get_info(self):
# print details in their own style
# Write a loop that iterates over objects of laptop and mobile and calls get_info() on each to show polymorphism
```

```python
# Parent Class
class Product:

    # Constructor
    def __init__(self, name, price, category):
        self.name = name
        self.__price = price   # Private attribute
        self.category = category

    # Getter method
    def get_price(self):
        return self.__price

    # Setter method
    def set_price(self, new_price):
        if new_price > 0:
            self.__price = new_price
            print("Price updated successfully")
        else:
            print("Price must be greater than 0")

    # Parent method
    def get_info(self):
        print("Product Name:", self.name)
        print("Price:", self.get_price())
        print("Category:", self.category)


# Child Class: Laptop
class Laptop(Product):

    # Override get_info()
    def get_info(self):
        print("----- Laptop Details -----")
        print("Laptop Name:", self.name)
        print("Laptop Price:", self.get_price())
        print("Laptop Category:", self.category)


# Child Class: Mobile
class Mobile(Product):

    # Override get_info()
    def get_info(self):
        print("----- Mobile Details -----")
        print("Mobile Name:", self.name)
        print("Mobile Price:", self.get_price())
        print("Mobile Category:", self.category)


# Create objects
laptop1 = Laptop("Dell Inspiron", 60000, "Electronics")
mobile1 = Mobile("Samsung Galaxy", 25000, "Electronics")


# Store objects in a list
products = [laptop1, mobile1]


# Loop through objects (Polymorphism)
for product in products:
    product.get_info()
    print()
```
