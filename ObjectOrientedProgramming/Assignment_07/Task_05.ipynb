## Abstruction(using Abstract base Class)

```python
# Create an abstract class Payment with abstract method:
# product_payment(amount)
# then create two subclasses:
# CreaditCardPayment
# UPIPayment
# Both override process_payment() simple print statements.
# test all classes.
```

```python
from abc import ABC, abstractmethod


# Abstract Class
class Payment(ABC):

    # Abstract Method
    @abstractmethod
    def process_payment(self, amount):
        pass


# Subclass: Credit Card Payment
class CreditCardPayment(Payment):

    # Override abstract method
    def process_payment(self, amount):
        print(f"Credit Card Payment of {amount} processed successfully.")


# Subclass: UPI Payment
class UPIPayment(Payment):

    # Override abstract method
    def process_payment(self, amount):
        print(f"UPI Payment of {amount} processed successfully.")


# Create objects
credit = CreditCardPayment()
upi = UPIPayment()

# Test all classes
credit.process_payment(5000)
upi.process_payment(1200)
```
