## Task 1: Basic Class & Object Creation

```python
# 1. Create a class Product with:
# 2. Attribute : name , price, category
# 3. Method: get_info() --> print product details
# 4. Create two objects and call get_info().
# 5. extra(optional) : Add a method apply_discount(percent) that return the discounted prices
```

```python
# Create a class Product
class Product:

    # Constructor
    def __init__(self, name, price, category):
        self.name = name
        self.price = price
        self.category = category

    # Method to print product details
    def get_info(self):
        print("Product Name:", self.name)
        print("Price:", self.price)
        print("Category:", self.category)

    # Optional method to apply discount
    def apply_discount(self, percent):
        discount_price = self.price - (self.price * percent / 100)
        return discount_price


# Create two objects
product1 = Product("Laptop", 50000, "Electronics")
product2 = Product("Shoes", 2000, "Fashion")

# Call get_info()
product1.get_info()
print()

product2.get_info()
print()

# Apply discount
print("Discounted Price of Laptop:", product1.apply_discount(10))
print("Discounted Price of Shoes:", product2.apply_discount(20))
```
