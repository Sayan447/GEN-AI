## Magic Methods & Operator Overloading

```python
# Add the following to your Product class:
# 1. __str__ method
# return a readble string:
# product(name , price , category)
# 2. Operator Overloading(__add__)
# Allow : 
# product1 + product2
# to return the total combined price
# test this with two product objects.
```

```python
# Create Product class
class Product:

    # Constructor
    def __init__(self, name, price, category):
        self.name = name
        self.__price = price   # Private attribute
        self.category = category

    # Getter method
    def get_price(self):
        return self.__price

    # Setter method
    def set_price(self, new_price):
        if new_price > 0:
            self.__price = new_price
            print("Price updated successfully")
        else:
            print("Price must be greater than 0")

    # Method to print product details
    def get_info(self):
        print("Product Name:", self.name)
        print("Price:", self.get_price())
        print("Category:", self.category)

    # Magic Method: __str__
    def __str__(self):
        return f"Product({self.name}, {self.__price}, {self.category})"

    # Operator Overloading: __add__
    def __add__(self, other):
        return self.__price + other.get_price()


# Create two product objects
product1 = Product("Laptop", 50000, "Electronics")
product2 = Product("Mobile", 25000, "Electronics")

# Test __str__
print(product1)
print(product2)

# Test __add__
total_price = product1 + product2
print("\nTotal Combined Price:", total_price)
```
