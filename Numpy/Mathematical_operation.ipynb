## Stacking and Spliting

##### Stacking means joining multiple arrays together.

##### Splitting means dividing an array into smaller arrays.

##### size of the array must be the same

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

result = np.vstack((a, b))
print(result)
```

```python
result = np.hstack((a, b))
print(result)
```
