## Numpy 

###### Numpy is a fundamental package for sciencetific computing  in python . It is a Python library that provides a multi dimentional array object and various derived object such as masked arrays or matrics and an assortment of routing for fast operation on array including mathematical , logical , shape manipulation , sorting, selecting , I/O , discrete fourier transform , basic linear algebra , stastical operation , Random simulation, etc.

### Numpy Arrays Vs Python Sequence
#### Numpy arrays have a fixed size at creating , unlike python lists which grow dynamically chaning the size of an n dimentional array will create a new array and delete the original.



#### The elements in a Numpy array are required to be of the same data type and thus will be of the same size in memory.



#### Numpy array facilitate advanced math and other types of operation on larger number of data . Typically , such operation are executed more efficiently with less code than is possible using python's built in sequences.
