## Mini Use Case: Sales Analysis

```python
# Given daily sales data:
# sales = np.array([1200 , 1500 , 900 ,2000 ,1800 ,1700 ,1600])
# Perform:
# 1. Total weekly sales 
# 2. Average daily sales 
# 3. Highest and lowest sales day 
# 4. Standard deviation of sales 
# 5. Identify days where sales were above average
```

```python
import numpy as np

# Given daily sales data
sales = np.array( [1500, 900, 2000, 1800, 1700, 1600])

# 1. Total weekly sales
total_sales = np.sum(sales)

# 2. Average daily sales
average_sales = np.mean(sales)

# 3. Highest and lowest sales day
highest_sales = np.max(sales)
lowest_sales = np.min(sales)

highest_day = np.argmax(sales) + 1  # Day number
lowest_day = np.argmin(sales) + 1   # Day number

# 4. Standard deviation of sales
std_sales = np.std(sales)

# 5. Days where sales were above average
above_average_days = np.where(sales > average_sales)[0] + 1

# Print results
print("Total Weekly Sales:", total_sales)
print("Average Daily Sales:", average_sales)
print("Highest Sales:", highest_sales, "on Day", highest_day)
print("Lowest Sales:", lowest_sales, "on Day", lowest_day)
print("Standard Deviation:", std_sales)
print("Days with Sales Above Average:", above_average_days)
```
