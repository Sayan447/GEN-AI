## Important Math Operation

```python
# Give two array:
# A  = np.array([10,20,30,40])
# B = np.array ([1,2,3,4])
# perform and print result for:
# 1. Addition (A+B)
# 1. Substraction (A-B)
# 1. Multiplication (A/B)
# 1. Division (A*B)
# 1. Power (A**2)

# Extra(optional) : Use NUmpy functions : np.add(),np.substract() for same operation
```

```python
import numpy as np

# Create arrays
A = np.array([10, 20, 30, 40])
B = np.array([1, 2, 3, 4])

# Arithmetic Operations
print("Addition (A + B):", A + B)
print("Subtraction (A - B):", A - B)
print("Multiplication (A * B):", A * B)
print("Division (A / B):", A / B)
print("Power (A ** 2):", A ** 2)

# Using NumPy functions
print("\nUsing NumPy Functions:")
print("np.add(A, B):", np.add(A, B))
print("np.subtract(A, B):", np.subtract(A, B))
print("np.multiply(A, B):", np.multiply(A, B))
print("np.divide(A, B):", np.divide(A, B))
print("np.power(A, 2):", np.power(A, 2))
```
