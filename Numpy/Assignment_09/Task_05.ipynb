## Statistical Operation(core focus)

```python
# Given:
# marks = np.array([78,85,90,66,72,88,95,60])
# calculate:
# Mean
# Median
# Variance
# Standard & Deviation
# Minimum & Maximum
# Range(max-min)
```

```python
import numpy as np

# Given array
marks = np.array([85, 90, 66, 72, 88, 95, 60])

# Mean
mean_value = np.mean(marks)

# Median
median_value = np.median(marks)

# Variance
variance_value = np.var(marks)

# Standard Deviation
std_value = np.std(marks)

# Minimum and Maximum
min_value = np.min(marks)
max_value = np.max(marks)

# Range (max - min)
range_value = np.ptp(marks)   # or max_value - min_value

# Print results
print("Mean:", mean_value)
print("Median:", median_value)
print("Variance:", variance_value)
print("Standard Deviation:", std_value)
print("Minimum:", min_value)
print("Maximum:", max_value)
print("Range:", range_value)
```
