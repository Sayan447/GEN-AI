## Percentile & Sorting


```python
# Using the same marks array:
# 1. Sort the array
# 2. find:
#  25th percentile
#  50th percentile
#  75th percentile

# count how many student scored above the average marks
```

```python
import numpy as np

# Given array
marks = np.array([85, 90, 66, 72, 88, 95, 60])

# 1. Sort the array
sorted_marks = np.sort(marks)

# Percentiles
p25 = np.percentile(marks, 25)
p50 = np.percentile(marks, 50)
p75 = np.percentile(marks, 75)

# Average marks
average = np.mean(marks)

# Count students scoring above average
above_average_count = np.sum(marks > average)

# Print results
print("Sorted Array:", sorted_marks)
print("25th Percentile:", p25)
print("50th Percentile (Median):", p50)
print("75th Percentile:", p75)
print("Average Marks:", average)
print("Students Scoring Above Average:", above_average_count)
```
