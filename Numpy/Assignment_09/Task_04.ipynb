## Task 4 : Aggregation Operation

```python
# given a 2D array:
# data = np.array([
    # [10,20,30],
    # [40,50,60],
    # [70,80,90]
# ])

# Find:
# 1. Row-wise sum
# 2. Column-wise sum
# 3. Minimum value
# 4. Maximum value
# 5. Overall mean
```

```python
import numpy as np

# Given 2D array
data = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

# 1. Row-wise sum
row_sum = np.sum(data, axis=1)

# 2. Column-wise sum
col_sum = np.sum(data, axis=0)

# 3. Minimum value
min_value = np.min(data)

# 4. Maximum value
max_value = np.max(data)

# 5. Overall mean
mean_value = np.mean(data)

# Print results
print("Row-wise Sum:", row_sum)
print("Column-wise Sum:", col_sum)
print("Minimum Value:", min_value)
print("Maximum Value:", max_value)
print("Overall Mean:", mean_value)
```
